```
# 5. VERIFICAR RESULTADOS
import os
def listar(pasta, n=0):
    ind = "  "*n
    if os.path.exists(pasta):
        for i in sorted(os.listdir(pasta)):
            p = os.path.join(pasta, i)
            if os.path.isdir(p):
                print(f"{ind}📂 {i}/")
                listar(p, n+1)
            else:
                print(f"{ind}📄 {i} ({os.path.getsize(p):,} bytes)")

print("📁 ARQUIVOS GERADOS:\n")
print("output/")
listar("output", 1)
```
